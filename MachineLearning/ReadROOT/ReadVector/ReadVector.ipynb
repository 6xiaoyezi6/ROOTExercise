```python
import numpy as np
import uproot

with uproot.open("VectorData.root") as root_file:
    tree = root_file["tree"]

    EventID = tree["EventID"].array(library="np")
    Position = tree["Position"].array(library="np")

x = np.array([p.member("fX") for p in Position])
y = np.array([p.member("fY") for p in Position])
z = np.array([p.member("fZ") for p in Position])

data = np.column_stack((EventID, x, y, z))

print(f"{'EventID':>8} {'X':>10} {'Y':>10} {'Z':>10}")

for EventID, x, y, z in data:
    print(f"{EventID:8.0f} {x:10.2f} {y:10.2f} {z:10.2f}")
```
