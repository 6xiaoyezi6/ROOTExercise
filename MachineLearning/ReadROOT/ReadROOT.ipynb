# 使用 uproot 读取 ROOT 数据

- **功能**：读取 `Data.root` 中 `tree` TTree 的 `x`、`y`、`noise` 分支，将其转换为三列 NumPy 数组并打印。
- **方法**：使用 `uproot.open` 打开 ROOT 文件，通过 `tree.arrays(..., library="np")` 读取指定分支，再用 `numpy.column_stack` 按列组合。
- **注意事项**：运行前需安装 `uproot` 和 `numpy`，并确保 `Data.root` 位于当前工作目录且包含名为 `tree` 的 TTree。

```python
import numpy as np
import uproot

with uproot.open("Data.root") as root_file:
    tree = root_file["tree"] #读取ROOT文件的树 TTree *tree = (TTree*)file->Get("tree");
    branches = tree.arrays(["x", "y", "noise"], library="np") #读取root文件中的三个branch，并转化为np结构的数据

data = np.column_stack((branches["x"], branches["y"], branches["noise"])) #column_stack 的意思是把三个branch的数据按列排布

print(f"{'x':>12} {'y':>12} {'noise':>12}")
for x, y, noise in data:
    print(f"{x:12.6f} {y:12.6f} {noise:12.6f}")
```

```python
import torch

# 把NumPy ndarray 转为适合于神经网络的 PyTorch Tensor 的数据结构
data_tensor = torch.tensor(data, dtype=torch.float32)

print(data_tensor)
print("shape =", data_tensor.shape)
print("dtype =", data_tensor.dtype)
```
